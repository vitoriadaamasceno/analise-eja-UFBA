```python
#Importações
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

## Bahia


```python
eleitorado_bahia = pd.read_csv('./cruzamento-bahia.csv')
display(eleitorado_bahia)
```

```python
#Escolaridades - Tipos
escolaridades= eleitorado_bahia['Escolaridade'].unique()
escolaridades
```

```python
# Dicionário para armazenar as somas
somas = {}

# Calculando a soma do eleitorado para cada escolaridade
for escolaridade in escolaridades:
    soma = eleitorado_bahia[eleitorado_bahia["Escolaridade"] == escolaridade]["Eleitorado"].sum()
    somas[escolaridade] = soma

# Criando uma nova tabela
somas_escolaridade_bahia = pd.DataFrame(somas.items(), columns=["Escolaridade", "Soma Eleitorado"])

somas_escolaridade_bahia
```

```python
#Gerando grafico do total de escolaridade
plt.figure(figsize=(10, 8))
bars = plt.bar(eleitorado_bahia["Escolaridade"], eleitorado_bahia["Eleitorado"],label='Total Eleitorado', color='green')

# Adicionando rótulos e título
plt.xlabel("Escolaridade")
plt.ylabel("Soma Eleitorado")
plt.title("Soma do Eleitorado da Bahia por Escolaridade")
plt.xticks(rotation=45, ha='right')

for bar in bars:
    yval = bar.get_height()
    plt.text(bar.get_x() + bar.get_width()/2, yval, round(yval, 2), ha='center', va='bottom', color='black', fontsize=10)

# Exibindo o gráfico
plt.legend(fontsize=10)
plt.tight_layout()
plt.show()
```

```python
eleitorado_faixa_etaria_bahia = pd.read_csv('./faixa-etaria-bahia.csv')
display(eleitorado_faixa_etaria_bahia)
```

### Analfabetos Bahia

```python
dados_analfabetos_bahia = eleitorado_faixa_etaria_bahia[eleitorado_faixa_etaria_bahia['Escolaridade'] == 'ANALFABETO']
# Reorganizar as linhas
first_item = dados_analfabetos_bahia.iloc[0]  # Primeira linha
dados_analfabetos_bahia =dados_analfabetos_bahia.drop(0)  # Remove a primeira linha
dados_analfabetos_bahia= pd.concat([dados_analfabetos_bahia, pd.DataFrame([first_item])], ignore_index=True)  
dados_analfabetos_bahia
```

```python

# Criar um gráfico dos dados dos analfabetos 
fig, ax = plt.subplots(figsize=(12, 8))
bars = ax.bar(dados_analfabetos_bahia['Faixa etária'],dados_analfabetos_bahia['Eleitorado'], label='Total Eleitorado', color='g')

#rótulos e título
ax.set_xlabel('Faixas Etárias')
ax.set_ylabel('Total Eleitorado')
ax.set_title('Distribuição do Total de Eleitores da Bahia por Faixa Etária (Escolaridade: ANALFABETO)')

# valores em cima de cada barra
for bar, value in zip(bars, dados_analfabetos_bahia['Eleitorado']):
    ax.text(bar.get_x() + bar.get_width() / 2, bar.get_height(), str(value), ha='center', va='bottom',fontsize=10)

plt.xticks(rotation=45, ha='right')

ax.legend(fontsize=10)

plt.tight_layout()
plt.show()
```
